- `Step one`, if we decide to add all those dummy variables to the current state of df,we would observe duplicate information that is the categorical variable Reason for Absence
on one side, would convey the same information as all 27 columns from reason_columns on the other. this phenomenon is called **multicollinearity** and it is something that should be avoided in general. Therefore, in our case, we could drop
- `Step two`, **grouping the variables**.If we add all these dummy variables in the df data frame,we would end up with a data set containing nearly 40 columns. This sounds like too much when dealing with 700 observations.
- so, group in qualitative reasons 

```python
# droping Reason for Absence columns as it is the same as reason_col
df = df.drop(['Reason for Absence'], axis=1)
```

```python
df
```

**Grouping Reason for Absence**
- `grp_1`: (1-14)diff diseases
- `grp_2`: (15-17)pregnancy related
- `grp_3`: (18-21)Poisoning or others
- `grp_4`: (22-28)light reasons for absence

```python
# now separating them for different Reason for Absence
reason_type1 = reason_col.loc[:,1:14].max(axis=1)
reason_type2 = reason_col.loc[:,15:17].max(axis=1)
reason_type3 = reason_col.loc[:,18:21].max(axis=1)
reason_type4 = reason_col.loc[:,22:].max(axis=1)
```

#### Concatenate Columns Values

```python
df = pd.concat([df, reason_type1, reason_type2, reason_type3, reason_type4], axis=1)
df
```
